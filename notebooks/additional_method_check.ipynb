```python
from algebraic_immunity_utils import Matrix
from restricted_algebraic_immunity.inductive_reed_muller.IV import IVRestrictedAI

def print_matrix(matrix):
    for row in matrix:
        print(" ".join(map(str, row)))
```

```python
m = Matrix([[1,1], [0,1]])
```

```python
m1 = m.copy()
support_slice = ['000', '010', '100']
monom_slice = ['000', '001', '010']
idx = 2
operations = [(0,1)]
n_m = m.compute_next(monom_slice, support_slice, idx, operations)
print_matrix(n_m.to_list())
print_matrix(m.to_list())
```

```python
m3 = IVRestrictedAI.compute_next(v_previous=m1,
                            support_slice=support_slice,
                            monom_slice=monom_slice, 
                            idx=idx,
                            operations=operations)
```
